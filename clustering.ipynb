```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris # iris dataseti
from sklearn.preprocessing import StandardScaler # ölçeklendirme

from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering

import warnings
warnings.filterwarnings('ignore')
```

```python
iris = load_iris()
data = iris.data
label = iris.target

print(f"Veri şekli: {data.shape}")
print(f"Etiket şekli: {label.shape}")
print(f"Öznitelik isimleri: {iris.feature_names}")
print(f"Gerçek tür isimleri: {iris.target_names}")
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

print("Orijinal veri (ilk 5 satır):")
print(data[:5])
print("\nÖlçeklendirilmiş veri (ilk 5 satır):")
print(scaled_data[:5])
```

```python
kmeans = KMeans(n_clusters=3, random_state=42 , n_init = 10)

kmeans_labels = kmeans.fit_predict(data)
print("K-Means (ölçeksiz) sonuçları:")
print(kmeans_labels)

kmeans_labels_scaled = kmeans.fit_predict(scaled_data)
print("K-Means (ölçekli) sonuçları:")
print(kmeans_labels_scaled)
```

```python
kmeans_crosstab = pd.crosstab(label, kmeans_labels)
kmeans_crosstab_scaled = pd.crosstab(label, kmeans_labels_scaled)

print("Sütunlar: K-Means'in bulduğu kümeler (0, 1, 2)")
print("Satırlar: Gerçek Türler (0: Setosa, 1: Versicolor, 2: Virginica)\n")

print("K-Means (ölçeksiz) Karşılaştırma Tablosu:")
print(kmeans_crosstab)

print("\nK-Means (ölçekli) Karşılaştırma Tablosu:")
print(kmeans_crosstab_scaled)
```

```python
agg_cluster = AgglomerativeClustering(n_clusters=3)

agg_labels = agg_cluster.fit_predict(data)
print("Hiyerarşik Kümeleme (ölçeksiz) sonuçları:")
print(agg_labels)

agg_labels_scaled = agg_cluster.fit_predict(scaled_data)
print("Hiyerarşik Kümeleme (ölçekli) sonuçları:")
print(agg_labels_scaled)
```

```python
agg_crosstab = pd.crosstab(label, agg_labels)
agg_crosstab_scaled = pd.crosstab(label, agg_labels_scaled)

print("Sütunlar: Hiyerarşik Kümeleme'nin bulduğu kümeler (0, 1, 2)")
print("Satırlar: Gerçek Türler (0: Setosa, 1: Versicolor, 2: Virginica)")

print("\nHiyerarşik Kümeleme (ölçeksiz) Karşılaştırma Tablosu:")
print(agg_crosstab)

print("\nHiyerarşik Kümeleme (ölçekli) Karşılaştırma Tablosu:")
print(agg_crosstab_scaled)
```
